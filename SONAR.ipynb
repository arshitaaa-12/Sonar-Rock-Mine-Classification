**Problem:** Classify sonar signals as either Rock or Mine using machine learning.

Importing the dependencies

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
```

Data Collection and Data Processing

```python
from google.colab import files
uploaded = files.upload()
```

```python
#loading the dataset to a pandas dataframe
sonar_data = pd.read_csv('Copy of sonar data.csv', header=None)
```

```python
sonar_data.head()
```

```python
#number of rows and column
print("Dataset shape: ", sonar_data.shape)
```

```python
#describe tells tha statistical measures of the data
sonar_data.describe()
```

```python
#60 is the column for rock and mine
sonar_data[60].value_counts()
```

M -> Mine

R -> Rock

```python
sonar_data.groupby(60).mean()
```

```python
#separating data and labels
X = sonar_data.drop(columns=60, axis=1)
Y = sonar_data[60]
```

```python
print(X)
print(Y)
```

```python
import matplotlib.pyplot as plt

# Class distribution
print("Class distribution:")
print(Y.value_counts())

Y.value_counts().plot(kind='bar')

plt.xlabel("Class")
plt.ylabel("Number of Samples")
plt.title("Rock vs Mine Distribution")
plt.show()
```

Training and Test Data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.1, stratify = Y, random_state = 1)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python
print(X_train)
print(Y_train)
```

Model Training -> Logistic Regression

```python
model = LogisticRegression()
```

```python
#training the logistic regression model with training data
model.fit(X_train, Y_train)
```

Model Evaluation

```python
#accuracy on training data
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction, Y_train)
```

```python
print('Accuracy on training data: ', training_data_accuracy)
```

```python
#accuracy on test data
X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction, Y_test)
```

```python
print('Accuracy on test data: ', test_data_accuracy)
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns

# Confusion Matrix
cm = confusion_matrix(Y_test, X_test_prediction)

print("Confusion Matrix:")
print(cm)

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```

```python
from sklearn.metrics import classification_report

print("Classification Report:")
print(classification_report(Y_test, X_test_prediction))
```
